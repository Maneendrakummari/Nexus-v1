```python
%scala

import org.apache.spark.sql.functions._

// Read Bronze products
val products = spark.table("retail.bronze.products")

// Business columns used to compare product records
val businessColumns = Seq(
  "product_id",
  "product_name",
  "category",
  "unit_price",
  "active"
)

// Count total rows and distinct product versions per ID
val duplicateAnalysis = products
  .groupBy("product_id")
  .agg(
    count("*").alias("total_count"),
    countDistinct(
      struct(businessColumns.map(col): _*)
    ).alias("distinct_versions")
  )
  .filter(col("total_count") > 1)

// Exact duplicates: multiple rows, but only one distinct version
val exactDuplicateIds = duplicateAnalysis
  .filter(col("distinct_versions") === 1)
  .select("product_id")

// Conflicting duplicates: multiple distinct versions
val conflictingDuplicateIds = duplicateAnalysis
  .filter(col("distinct_versions") > 1)
  .select("product_id")

println("========== DUPLICATE ANALYSIS ==========")

println("Exact duplicate IDs:")
exactDuplicateIds.show(false)

println("Conflicting duplicate IDs:")
conflictingDuplicateIds.show(false)
```

```python
%scala

// 1. Remove exact duplicates
val deduplicatedProducts = products
  .dropDuplicates(businessColumns)

// 2. Get all records belonging to conflicting IDs
val conflictingProducts = products
  .join(
    conflictingDuplicateIds,
    Seq("product_id"),
    "inner"
  )

// 3. Remove conflicting IDs from the working dataset
val productsAfterDuplicateHandling = deduplicatedProducts
  .join(
    conflictingDuplicateIds,
    Seq("product_id"),
    "left_anti"
  )

println("========== DUPLICATE HANDLING ==========")

println(s"Original records: ${products.count()}")
println(s"After exact deduplication: ${deduplicatedProducts.count()}")
println(s"Conflicting records: ${conflictingProducts.count()}")
println(s"Records after duplicate handling: ${productsAfterDuplicateHandling.count()}")

println("========== CONFLICTING PRODUCTS ==========")
conflictingProducts.select(businessColumns.map(col): _*).show(false)
```

```python
%scala

import org.apache.spark.sql.functions._

// Convert price safely to decimal
val validatedProducts = productsAfterDuplicateHandling
  .withColumn(
    "unit_price",
    expr("try_cast(unit_price AS DECIMAL(10,2))")
  )
  .withColumn(
    "category",
    initcap(trim(col("category")))
  )
  .withColumn(
    "product_name",
    trim(col("product_name"))
  )

// Build rejection reasons
val productsWithReasons = validatedProducts
  .withColumn(
    "rejection_reason",
    concat_ws(
      ", ",
      when(
        col("product_id").isNull || trim(col("product_id")) === "",
        lit("MISSING_PRODUCT_ID")
      ),
      when(
        col("product_name").isNull || col("product_name") === "",
        lit("MISSING_PRODUCT_NAME")
      ),
      when(
        col("category").isNull || col("category") === "",
        lit("MISSING_CATEGORY")
      ),
      when(
        col("unit_price").isNull,
        lit("INVALID_OR_MISSING_PRICE")
      ).when(
        col("unit_price") <= 0,
        lit("NON_POSITIVE_PRICE")
      ),
      when(
        col("active").isNull ||
        !upper(trim(col("active"))).isin("Y", "N"),
        lit("INVALID_ACTIVE_FLAG")
      )
    )
  )

// Separate valid and rejected records
val validProducts = productsWithReasons
  .filter(col("rejection_reason") === "")

val rejectedProducts = productsWithReasons
  .filter(col("rejection_reason") =!= "")

println("========== VALIDATION RESULTS ==========")
println(s"Records before validation: ${productsAfterDuplicateHandling.count()}")
println(s"Valid records: ${validProducts.count()}")
println(s"Rejected records: ${rejectedProducts.count()}")

println("========== REJECTED PRODUCTS ==========")
rejectedProducts
  .select("product_id", "product_name", "category", "unit_price", "rejection_reason")
  .show(false)

println("========== VALID PRODUCTS ==========")
validProducts
  .select("product_id", "product_name", "category", "unit_price", "active")
  .show(false)
```

```python
%scala

import org.apache.spark.sql.functions._

// Add conflicting records to rejected dataset
val conflictingWithReason = conflictingProducts
  .withColumn(
    "unit_price",
    expr("try_cast(unit_price AS DECIMAL(10,2))")
  )
  .withColumn(
    "category",
    initcap(trim(col("category")))
  )
  .withColumn(
    "product_name",
    trim(col("product_name"))
  )
  .withColumn(
    "rejection_reason",
    lit("CONFLICTING_PRODUCT_ID_DIFFERENT_PRICE")
  )

// Combine all rejected records
val allRejectedProducts = rejectedProducts
  .unionByName(conflictingWithReason)

// Write valid products to Silver
validProducts
  .drop("rejection_reason")
  .write
  .format("delta")
  .mode("overwrite")
  .saveAsTable("retail.silver.products")

// Write rejected products to Silver
allRejectedProducts
  .write
  .format("delta")
  .mode("overwrite")
  .saveAsTable("retail.silver.rejected_products")

println("========== SILVER TABLES CREATED ==========")

spark.sql("SHOW TABLES IN retail.silver").show(false)

println("========== SILVER PRODUCTS ==========")
spark.table("retail.silver.products").show(false)

println("========== REJECTED PRODUCTS ==========")
spark.table("retail.silver.rejected_products")
  .select("product_id", "rejection_reason")
  .show(false)
```
