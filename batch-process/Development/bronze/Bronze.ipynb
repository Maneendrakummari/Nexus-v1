```python
%scala
val productsPath =
  "abfss://raw@adlsacntext.dfs.core.windows.net/nexus/products.csv"

val bronzePath =
  "abfss://raw@adlsacntext.dfs.core.windows.net/project_nexus/bronze/products"
```

```python
%scala
val productsDF = spark.read
  .option("header", "true")
  .option("inferSchema", "false")
  .csv(productsPath)
```

```python
%scala
productsDF.show(false)

productsDF.printSchema()

println("Total records: " + productsDF.count())
```

```python
%scala
import org.apache.spark.sql.functions._

val bronzeProductsDF = productsDF
  .withColumn("_ingestion_timestamp", current_timestamp())
  .withColumn("_source_file", input_file_name())
  .withColumn("_batch_id", lit("BATCH_001"))
```

```python
%scala
bronzeProductsDF.write
  .format("delta")
  .mode("overwrite")
  .save(bronzePath)
```

```python
%scala

spark.sql("""
  CREATE TABLE IF NOT EXISTS retail.bronze.products
  USING DELTA
  LOCATION 'abfss://raw@adlsacntext.dfs.core.windows.net/project_nexus/bronze/products'
""")
```

```python
%scala
val df = spark.table("retail.bronze.products")

df.show(false)
df.printSchema()
println("Total records: " + df.count())
```
