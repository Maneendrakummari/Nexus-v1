```python
%scala
import org.apache.spark.sql.functions._
import org.apache.spark.sql.expressions.Window

// Read existing Silver orders
val existingOrders = spark.table("retail.silver.orders")

// Read incoming Bronze updates
val incomingUpdates = spark.table("retail.bronze.orders_updates")

// Convert timestamps
val existingWithTimestamp = existingOrders
  .withColumn(
    "_existing_updated_ts",
    to_timestamp(col("last_updated"), "yyyy-MM-dd HH:mm:ss")
  )

val incomingWithTimestamp = incomingUpdates
  .withColumn(
    "_incoming_updated_ts",
    to_timestamp(col("last_updated"), "yyyy-MM-dd HH:mm:ss")
  )

// Compare incoming updates with existing records
val updateComparison = incomingWithTimestamp
  .join(
    existingWithTimestamp.select(
      col("order_id"),
      col("_existing_updated_ts")
    ),
    Seq("order_id"),
    "left"
  )
  .withColumn(
    "record_type",
    when(col("_existing_updated_ts").isNull, "NEW_ORDER")
      .when(
        col("_incoming_updated_ts") > col("_existing_updated_ts"),
        "NEWER_UPDATE"
      )
      .when(
        col("_incoming_updated_ts") < col("_existing_updated_ts"),
        "STALE_UPDATE"
      )
      .otherwise("SAME_TIMESTAMP")
  )

// Display comparison
updateComparison.select(
  "order_id",
  "last_updated",
  "_existing_updated_ts",
  "record_type"
).orderBy("order_id").show(30, false)

// Summary
updateComparison.groupBy("record_type")
  .count()
  .show(false)
```

```python
%scala
import org.apache.spark.sql.functions._
import org.apache.spark.sql.expressions.Window

// Read original Bronze orders
val originalOrders = spark.table("retail.bronze.orders")

// Remove exact duplicates
val originalWithoutDuplicates = originalOrders
  .dropDuplicates(Seq(
    "order_id",
    "customer_id",
    "product_id",
    "quantity",
    "order_date",
    "order_status",
    "last_updated"
  ))

// Parse timestamps
val originalWithTimestamp = originalWithoutDuplicates
  .withColumn(
    "_original_updated_ts",
    to_timestamp(col("last_updated"), "yyyy-MM-dd HH:mm:ss")
  )

// Select latest original version for each order
val originalWindow = Window
  .partitionBy("order_id")
  .orderBy(col("_original_updated_ts").desc)

val originalLatest = originalWithTimestamp
  .withColumn("_rank", row_number().over(originalWindow))
  .filter(col("_rank") === 1)
  .select("order_id", "_original_updated_ts")

// Compare incoming updates with original Bronze data
val correctedComparison = incomingWithTimestamp
  .join(originalLatest, Seq("order_id"), "left")
  .withColumn(
    "record_type",
    when(col("_original_updated_ts").isNull, "NEW_ORDER")
      .when(
        col("_incoming_updated_ts") > col("_original_updated_ts"),
        "CORRECTION"
      )
      .when(
        col("_incoming_updated_ts") < col("_original_updated_ts"),
        "STALE_UPDATE"
      )
      .otherwise("SAME_TIMESTAMP")
  )

correctedComparison.select(
  "order_id",
  "last_updated",
  "_original_updated_ts",
  "record_type"
).orderBy("order_id").show(30, false)

correctedComparison.groupBy("record_type")
  .count()
  .show(false)
```

```python
%scala

import org.apache.spark.sql.functions._

// Valid reference IDs from Silver
val validCustomerIds = spark.table("retail.silver.customers")
  .select("customer_id")

val validProductIds = spark.table("retail.silver.products")
  .select("product_id")

val validStatuses = Seq(
  "PLACED", "SHIPPED", "DELIVERED",
  "CANCELLED", "RETURNED"
)

// Normalize incoming updates
val normalizedUpdates = incomingUpdates
  .withColumn("quantity", expr("try_cast(quantity AS INT)"))
  .withColumn(
    "order_date",
    coalesce(
      expr("try_to_date(trim(order_date), 'yyyy-MM-dd')"),
      expr("try_to_date(trim(order_date), 'dd-MM-yyyy')")
    )
  )
  .withColumn("order_status", upper(trim(col("order_status"))))
  .withColumn(
    "last_updated_ts",
    to_timestamp(col("last_updated"), "yyyy-MM-dd HH:mm:ss")
  )

// Join with valid customer and product references
val updatesWithReferences = normalizedUpdates
  .join(validCustomerIds, Seq("customer_id"), "left")
  .join(validProductIds, Seq("product_id"), "left")

// Add rejection reasons
val validatedUpdates = updatesWithReferences
  .withColumn(
    "rejection_reason",
    concat_ws(" | ",
      when(col("quantity").isNull || col("quantity") <= 0,
        lit("INVALID_QUANTITY")),
      when(col("order_date").isNull,
        lit("INVALID_ORDER_DATE")),
      when(col("order_date") > lit("2026-11-20").cast("date"),
        lit("FUTURE_ORDER_DATE")),
      when(col("order_status").isNull ||
        !col("order_status").isin(validStatuses: _*),
        lit("INVALID_ORDER_STATUS")),
      when(col("last_updated_ts").isNull,
        lit("INVALID_LAST_UPDATED")),
      when(col("customer_id").isNull ||
        trim(col("customer_id")) === "",
        lit("MISSING_CUSTOMER_ID")),
      when(col("customer_id").isNotNull &&
        trim(col("customer_id")) =!= "" &&
        validCustomerIds.select("customer_id")
          .where(col("customer_id").isNotNull).count() >= 0 &&
        col("customer_id").isNotNull,
        lit("CHECK_CUSTOMER_REFERENCE"))
    )
  )
```

```python
%scala
// Rename reference columns
val customerReferenceV2 = validCustomerIds
  .withColumnRenamed("customer_id", "_matched_customer_id")

val productReferenceV2 = validProductIds
  .withColumnRenamed("product_id", "_matched_product_id")

// Join with valid references
val updatesWithReferencesV2 = normalizedUpdates
  .join(
    customerReferenceV2,
    col("customer_id") === col("_matched_customer_id"),
    "left"
  )
  .join(
    productReferenceV2,
    col("product_id") === col("_matched_product_id"),
    "left"
  )

// Validate records
val validatedUpdatesV2 = updatesWithReferencesV2
  .withColumn(
    "rejection_reason",
    concat_ws(" | ",
      when(col("quantity").isNull || col("quantity") <= 0,
        lit("INVALID_QUANTITY")),
      when(col("order_date").isNull,
        lit("INVALID_ORDER_DATE")),
      when(col("order_date") > lit("2026-11-20").cast("date"),
        lit("FUTURE_ORDER_DATE")),
      when(col("order_status").isNull ||
        !col("order_status").isin(validStatuses: _*),
        lit("INVALID_ORDER_STATUS")),
      when(col("last_updated_ts").isNull,
        lit("INVALID_LAST_UPDATED")),
      when(col("customer_id").isNull ||
        trim(col("customer_id")) === "",
        lit("MISSING_CUSTOMER_ID")),
      when(col("customer_id").isNotNull &&
        trim(col("customer_id")) =!= "" &&
        col("_matched_customer_id").isNull,
        lit("INVALID_CUSTOMER_REFERENCE")),
      when(col("product_id").isNull ||
        trim(col("product_id")) === "",
        lit("MISSING_PRODUCT_ID")),
      when(col("product_id").isNotNull &&
        trim(col("product_id")) =!= "" &&
        col("_matched_product_id").isNull,
        lit("INVALID_PRODUCT_REFERENCE"))
    )
  )
  .drop("_matched_customer_id", "_matched_product_id")

// Separate valid and rejected records
val validIncomingUpdatesV2 = validatedUpdatesV2
  .filter(col("rejection_reason") === "")

val rejectedIncomingUpdatesV2 = validatedUpdatesV2
  .filter(col("rejection_reason") =!= "")

// Display results
println("Total incoming: " + validatedUpdatesV2.count())
println("Valid: " + validIncomingUpdatesV2.count())
println("Rejected: " + rejectedIncomingUpdatesV2.count())

rejectedIncomingUpdatesV2.select(
  "order_id",
  "quantity",
  "customer_id",
  "product_id",
  "rejection_reason"
).show(false)
```

```python
%scala
// Keep only classification information
val updateTypes = correctedComparison
  .select("order_id", "record_type")

// Combine validation and classification
val classifiedValidUpdates = validIncomingUpdatesV2
  .join(updateTypes, Seq("order_id"), "inner")

// Select only eligible records
val mergeSource = classifiedValidUpdates
  .filter(
    col("record_type").isin("CORRECTION", "NEW_ORDER")
  )
  .drop("record_type", "last_updated_ts", "rejection_reason")

// Check counts
println("Records prepared for MERGE: " + mergeSource.count())

println("Corrections: " +
  mergeSource.filter(
    col("order_id").isin(
      "ORD000011", "ORD000012", "ORD000013",
      "ORD000014", "ORD000015", "ORD000016",
      "ORD000021", "ORD000031"
    )
  ).count())

println("New orders: " +
  mergeSource.filter(col("order_id").like("ORD00015%")).count())

// Check duplicate IDs
println("Duplicate IDs in MERGE source: " +
  mergeSource.groupBy("order_id")
    .count()
    .filter(col("count") > 1)
    .count())

mergeSource.select(
  "order_id", "customer_id", "product_id",
  "quantity", "order_status", "last_updated"
).orderBy("order_id").show(false)
```

```python
%scala

import io.delta.tables.DeltaTable

// Load the existing Silver Delta table
val silverOrders = DeltaTable.forName(
  spark,
  "retail.silver.orders"
)

// Execute MERGE
silverOrders.as("target")
  .merge(
    mergeSource.as("source"),
    "target.order_id = source.order_id"
  )
  .whenMatched(
    """
    to_timestamp(source.last_updated, 'yyyy-MM-dd HH:mm:ss')
    >
    to_timestamp(target.last_updated, 'yyyy-MM-dd HH:mm:ss')
    """
  )
  .updateAll()
  .whenNotMatched()
  .insertAll()
  .execute()

println("Delta MERGE completed successfully!")
```

```python
%scala
val incomingBatchId = "BATCH_002"

val existingBatch = spark.table("retail.silver.batch_control")
  .filter(col("batch_id") === incomingBatchId)

if (existingBatch.count() > 0) {

  val status = existingBatch
    .select("batch_status")
    .first()
    .getString(0)

  if (status == "SUCCESS") {
    println("Batch already processed successfully. Skipping execution.")

  } else if (status == "FAILED") {
    println("Previous execution failed. Retry is allowed.")

  } else {
    println("Batch is already in progress. Check before retrying.")
  }

} else {
  println("New batch. Processing can start.")
}
```
